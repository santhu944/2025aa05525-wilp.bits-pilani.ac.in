```python
G = 222  
random.seed(G)
np.random.seed(G)
```

```python
K = (G % 3) + 5

print("Group Number =", G)
print("Number of Medicines =", K)
```

```python
hidden_probs = []

for i in range(K):
    p = 0.4 + ((G + i) % 6) * 0.07
    hidden_probs.append(round(p,3))

print(hidden_probs)
```

```python
patients = []

for patient_id in range(1000):

    severity = (patient_id % 5) + 1

    patients.append({
        "patient_id": patient_id,
        "severity_score": severity
    })

df = pd.DataFrame(patients)
```
